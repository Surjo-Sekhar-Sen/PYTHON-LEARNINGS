```python
# MOUNTAIN PATTERN
n = int(input("How many rows you want: "))

for i in range(1, n + 1):
    for j in range(1, (n - i) + 1):
        print(" ", end = " ")
    for k in range(1, i + 1):
        print("*", end = " ")
    for l in range(1, (i - 1) + 1):
        print("*", end = " ")
    print(end = "\n")
```

```python
# REVERSE MOUNTAIN PATTERN
n = int(input("How many rows you want to print: "))

for i in range(1, n + 1):
    for k in range(1, (i-1)+1):
        print(" ", end = " ")
    for j in range(i, n + 1):
        print("*", end = " ")
    for l in range(i, (n - 1) + 1):
        print("*", end = " ")
    print(end = "\n")
```

```python
# PYRAMID PATTERN
n = int(input("How many rows you want to print:"))

for i in range(1, n + 1):
    for k in range(1, (n - i) + 1):
        print(" ", end = " ")
    for j in range(1, (i + 1)):
        print(" * ", end = " ")
    print(end = "\n")
```

```python
# BUTTERFLY PATTERN
n = int(input("How many rows you want to print:"))

for i in range(1, n + 1):
    for j in range(1, i + 1):
        print("*", end = " ")
    for k in range(1, (2*n - 2*i - 1) + 1): #earlier -> (n - i) + 1
        print(" ", end = " ")
    # for l in range(1, (n - i - 1) + 1):
    #     print(" ", end = " ")
    for m in range(1, i + 1):
        if m == n:
            print("", end = " ")
        else:
            print("*", end = " ")
    print(end = "\n")
for a in range(1, (n - 1) + 1):
    for b in range(1, (n - a) + 1):
        print("*", end = " ")
    for c in range(1, (2*a - 1) + 1): # EARLIER -> (a + 1)
        print(" ", end = " ")
    # for d in range(1, (a - 1) + 1):
    #     print(" ", end = " ")
    for e in range(1, (n - a) + 1):
        print("*", end = " ")
    print(end = "\n")
```
