```python
#GUESSING GAME REVISION
import random
random_num = random.randint(1, 101)
count = 0
```

```python
num = int(input("Guess the number: "))
while num != random_num:
    if num > random_num:
        print("Guess Lower")
        count = count + 1
        num = int(input("Guess again: "))
        continue
    else:
        print("Guess Higher")
        count = count + 1
        num = int(input("Guess again: "))
        continue
if num == random_num:
    count = count + 1
    print("Yayy!! You guessed it correct")
    if count == 1:
        print("You guessed it in: ", count, " try!!")
    else:
        print("You guessed it in: ", count, " tries!!")
```
